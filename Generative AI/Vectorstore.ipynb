```python
from langchain.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter  
```

```python
loader = TextLoader("speech.txt")
docums = loader.load()
```

```python
splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=30)
docs = splitter.split_documents(docums)
```

```python
docs
```

```python
embedd = OllamaEmbeddings(model="gemma2:2b")
db = FAISS.from_documents(docs, embedd)
```

```python
db
```

```python
query = "How many types of speech are there?"
ress = db.similarity_search(query)
ress
```

```python
ress[0].page_content
```

### Retriever

```python
retriever = db.as_retriever()
```

```python
retriever.invoke(query)
```

### Similarity seach with score

```python
scoring = db.similarity_search_with_score(query)
```

```python
scoring
```
